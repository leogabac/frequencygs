# Collective horizontal pathway

All horizontal particles move together; vertical particles remain in their initial positions. These figures supplement the earlier vertical scans. The coordinate starts at $s_h=5\,\mu$m and crosses the hill at zero; $\alpha=(5-s_h)/10$. We integrate $\gamma\dot s_h=-\partial_{s_h}(U_{\mathrm{total}}/100)$ using the source mobility, without fitting a time factor.

```python
%matplotlib inline
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
repo = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src/py/collective_horizontal_analysis.py').exists())
sys.path.insert(0, str(repo/'src/py'))
import collective_horizontal_analysis as h
out = h.OUT
```

```python
with np.load(out/'collective_profiles.npz') as data:
    profiles = pd.DataFrame(data['values'], columns=data['columns'])
equilibrium = pd.read_csv(out/'equilibrium_extrema.csv')
paths = pd.read_csv(out/'horizontal_dynamics.csv')
summary = pd.read_csv(out/'switching_summary.csv')
display(summary[['tau_s', 'omega_rad_s', 'crossing_angle_deg', 'observed_cold_half_crossing_angle_deg', 'angle_difference_deg']])
```

The left panel moves all horizontals, counting each pair once. The right panel shows static extrema; the dynamics below retain the lag behind that static minimum.

```python
h.plot_potential(profiles, equilibrium)
plt.show()
```

The dashed line marks hill crossing. Compare the calculated crossing angle with the halfway marker from the saved cold crossing fraction. At omega=0.19635 rad/s the angles are 51.56 and 51.75 degrees. No model parameters were fitted to these markers.

```python
h.plot_dynamics(paths, summary)
plt.show()
```
